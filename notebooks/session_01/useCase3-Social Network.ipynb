# Use Case 4: Social Network (Multiple Relationship Types)

Two people can be friends, colleagues, and family members simultaneously.

```python
import networkx as nx

S = nx.MultiGraph(name="Social Network")

people = ["Alice", "Bob", "Carol", "David", "Eve"]
S.add_nodes_from(people)

relationships = [
    ("Alice", "Bob",   {"type": "friend",    "since": 2015, "strength": 0.9}),
    ("Alice", "Bob",   {"type": "colleague", "since": 2020, "strength": 0.6}),
    ("Bob",   "Carol", {"type": "friend",    "since": 2018, "strength": 0.8}),
    ("Bob",   "Carol", {"type": "family",    "since": 1990, "strength": 1.0}),
    ("Carol", "David", {"type": "colleague", "since": 2022, "strength": 0.5}),
    ("Alice", "Eve",   {"type": "friend",    "since": 2019, "strength": 0.7}),
    ("Alice", "Alice", {"type": "self",      "since": 0,    "strength": 1.0}),  # self-loop
]
S.add_edges_from(relationships)
```

## Query1: Find all relationship types in the network

```python
rel_types = {d["type"] for _, _, d in S.edges(data=True)}
print("Relationship types:", rel_types)
```

## Query2: Find all relationships between Alice and Bob

```python
print("\nAlice ↔ Bob relationships:")
for key, data in S["Alice"]["Bob"].items():
    print(f"  [{key}] {data['type']} since {data['since']}")
```

## Query3: Average friendship strength

```python
friendships = [(u, v, d["strength"]) for u, v, d in S.edges(data=True)
               if d.get("type") == "friend"]
avg = sum(s for *_, s in friendships) / len(friendships)
print(f"\nAverage friendship strength: {avg:.2f}")
```
