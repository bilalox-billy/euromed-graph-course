# Use Case 2: Financial Transaction Network

Banks can perform multiple transactions to the same counterparty. Each transaction is an independent edge.

```python
import networkx as nx
from datetime import datetime

F = nx.MultiGraph(name="Interbank Transactions")

# Represent transactions as edges
transactions = [
    ("BankA", "BankB", {"amount": 1_000_000, "date": "2026-09-01", "type": "SWIFT"}),
    ("BankA", "BankB", {"amount":   500_000, "date": "2026-09-05", "type": "ACH"}),
    ("BankA", "BankB", {"amount": 2_000_000, "date": "2026-09-10", "type": "SWIFT"}),
    ("BankB", "BankC", {"amount":   750_000, "date": "2026-09-02", "type": "SEPA"}),
    ("BankA", "BankC", {"amount": 1_200_000, "date": "2026-09-03", "type": "SWIFT"}),
    ("BankA", "BankA", {"amount":   300_000, "date": "2026-09-07", "type": "internal"}),
]
F.add_edges_from(transactions)


```

## Query1: Number of transactions per pair

```python

print(f"BankA ↔ BankB transactions: {F.number_of_edges('BankA', 'BankB')}")

```

## Query2: SWIFT transactions only

```python
swift = [(u, v, k, d) for u, v, k, d in F.edges(keys=True, data=True)
         if d.get("type") == "SWIFT"]
print(f"\nSWIFT transactions:")
for u, v, k, d in swift:
    print(f"  {u} ↔ {v}: ${d['amount']:,} on {d['date']}")
```
