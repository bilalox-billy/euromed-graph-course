```python
# ─────────────────────────────────────────────────────────────────────────────
# Nodes and edges can carry any extra information as key-value attributes.
# This is how you enrich a graph with real-world data.
# ─────────────────────────────────────────────────────────────────────────────

import networkx as nx
P = nx.Graph()

# Add nodes with attributes (passed as keyword arguments)
P.add_node("Alice", age=30, city="London",    role="engineer")
P.add_node("Bob",   age=25, city="Manchester", role="designer")
P.add_node("Carol", age=35, city="London",    role="manager")

# Add edges with attributes
P.add_edge("Alice", "Bob",   relation="colleague", years_known=3)
P.add_edge("Alice", "Carol", relation="friend",    years_known=8)

# ─────────────────────────────────────────────────────────────────────────────
# Inspect node attributes
# G.nodes[node] returns a dictionary of that node's attributes
# ─────────────────────────────────────────────────────────────────────────────
print("=" * 50)
print("NODE ATTRIBUTES")
print("=" * 50)

print("Alice's attributes:", P.nodes["Alice"])
# → {'age': 30, 'city': 'London', 'role': 'engineer'}

print("\nAll nodes with attributes:")
for node, attrs in P.nodes(data=True):
    print(f"  {node}: {attrs}")

# ─────────────────────────────────────────────────────────────────────────────
# Inspect edge attributes
# ─────────────────────────────────────────────────────────────────────────────
print("\n" + "=" * 50)
print("EDGE ATTRIBUTES")
print("=" * 50)

print("Alice–Bob edge:", P.edges["Alice", "Bob"])
# → {'relation': 'colleague', 'years_known': 3}

print("\nAll edges with attributes:")
for u, v, attrs in P.edges(data=True):
    print(f"  {u} ── {v}: {attrs}")

# ─────────────────────────────────────────────────────────────────────────────
# Update an attribute on the fly
# ─────────────────────────────────────────────────────────────────────────────
P.nodes["Alice"]["city"] = "Edinburgh"   # Alice moved cities
print("\nAlice's updated city:", P.nodes["Alice"]["city"])
```
