## Paths, Cycles, and Connected Componnents

```python
# ─────────────────────────────────────────────────────────────────────────────
# A graph with TWO separate connected components (two isolated islands)
#
#  Component 1: A -- B -- C
#  Component 2: D -- E
#
# There is NO path from any node in component 1 to any node in component 2.
# ─────────────────────────────────────────────────────────────────────────────

import networkx as nx
H = nx.Graph()
H.add_edges_from([("A", "B"), ("B", "C")])  # Island 1
H.add_edges_from([("D", "E")])              # Island 2

print("=" * 55)
print("CONNECTED COMPONENTS")
print("=" * 55)

# nx.connected_components() returns a generator of sets, each set being one component
components = list(nx.connected_components(H))
print(f"Number of connected components : {len(components)}")
# → 2 — our graph is split into two separate islands

for i, component in enumerate(components, start=1):
    print(f"  Component {i}: {component}")

# Is the graph connected? (can you reach every node from every other node?)
print(f"\nIs the graph connected? {nx.is_connected(H)}")
# → False — there are 2 components

```

```python
# ─────────────────────────────────────────────────────────────────────────────
# SHORTEST PATH between two nodes
# ─────────────────────────────────────────────────────────────────────────────
print("\n" + "=" * 55)
print("SHORTEST PATH")
print("=" * 55)

path = nx.shortest_path(H, source="A", target="C")
print(f"Shortest path from A to C : {path}")
# → [A, B, C] — we must go through B

path_length = nx.shortest_path_length(H, source="A", target="C")
print(f"Length of that path        : {path_length} edges")
# → 2 — two edges to traverse

# What happens if we try to find a path between separate components?
print("\nTrying to find a path from A to D (different components)...")
try:
    nx.shortest_path(H, source="A", target="D")
except nx.NetworkXNoPath as error:
    print(f"  Error caught: {error}")
    print("  Explanation: D is in a different component — no path exists.")
```

### Key Insight: Disconnected Graphs

When a graph has **multiple connected components**, many global statistics (like average shortest path length or diameter) **cannot be computed** for the whole graph directly — they are only meaningful within each component.

This is a very common pitfall for beginners. Always check `nx.is_connected(G)` before computing path-based statistics.
