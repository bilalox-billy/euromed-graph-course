# Use Case 1: Public Transport Network

Two cities can be connected by multiple transport modes (bus, train, flight). Each edge represents one connection with its own schedule and price.

## Build the Graph

```python
import networkx as nx

T = nx.MultiGraph(name="Transport Network")

# Add cities
cities = ["Paris", "Lyon", "Marseille", "Bordeaux"]
T.add_nodes_from(cities)

# Multiple connections between same cities
T.add_edge("Paris", "Lyon", key="TGV_001", mode="TGV",  duration_h=2.0, price_eur=45)
T.add_edge("Paris", "Lyon", key="TGV_002", mode="TGV",  duration_h=2.0, price_eur=65)
T.add_edge("Paris", "Lyon", key="BUS_A",   mode="bus",  duration_h=4.5, price_eur=15)
T.add_edge("Paris", "Lyon", key="AIR_1",   mode="flight",duration_h=1.2,price_eur=99)

T.add_edge("Lyon", "Marseille", key="TGV_010", mode="TGV", duration_h=1.75, price_eur=35)
T.add_edge("Lyon", "Marseille", key="BUS_B",   mode="bus", duration_h=3.5,  price_eur=12)

T.add_edge("Paris", "Bordeaux", key="TGV_020", mode="TGV",  duration_h=2.1, price_eur=55)
T.add_edge("Paris", "Bordeaux", key="AIR_2",   mode="flight",duration_h=1.0, price_eur=79)
```

## Query1: Find the cheapest option between Paris and Lyon

```python
paris_lyon_edges = T["Paris"]["Lyon"]
cheapest_key = min(paris_lyon_edges, key=lambda k: paris_lyon_edges[k]["price_eur"])
cheapest = paris_lyon_edges[cheapest_key]
print(f"Cheapest Paris→Lyon: {cheapest['mode']} at €{cheapest['price_eur']}")
```

## Query2: Count Connections between two cities

```python
print(f"Paris–Lyon connections: {T.number_of_edges('Paris', 'Lyon')}")
```

## Query3: List all modes available in the whole network

```python
modes = {d["mode"] for _, _, d in T.edges(data=True)}
print(f"Transport modes: {modes}")
```
